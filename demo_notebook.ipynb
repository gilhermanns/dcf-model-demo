# Perfect DCF Model Demo ✨
This notebook demonstrates the power of the `modern-automated-dcf` library by performing a full valuation of AAPL, MSFT, and TSLA.

```python
import sys
import os
import pandas as pd

# Add library path
sys.path.append(os.path.abspath("../modern-automated-dcf"))
from automated_dcf import DCFModel

tickers = ['AAPL', 'MSFT', 'TSLA']
results_list = []

for ticker in tickers:
    print(f"Valuing {ticker}...")
    model = DCFModel(ticker)
    model.fetch_data()
    res = model.run_dcf(years=10)
    results_list.append(res)
    
    # Export results
    model.export_to_excel(f"outputs/{ticker}_DCF_Model.xlsx")
    model.plot_all_charts(output_dir="outputs/charts")

print("Done!")
```

```python
# Display Summary
summary_df = pd.DataFrame([{
    'Ticker': r['ticker'],
    'Current Price': r['current_price'],
    'Implied Price': r['intrinsic_value'],
    'Upside (%)': r['upside'] * 100
} for r in results_list])
summary_df
```
